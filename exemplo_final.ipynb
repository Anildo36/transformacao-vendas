# Projeto Final - Limpeza e Tratamento de Dados

## Contexto

Uma empresa de e-commerce disponibilizou um arquivo contendo pedidos realizados durante os meses de abril a junho de 2025.

Antes de realizar qualquer análise comercial, a equipe de dados precisa preparar essa base.

Porém, o arquivo apresenta problemas comuns encontrados em dados reais:

 - Clientes sem identificação
 - Cidades escritas em formatos diferentes
 - Status com letras maiúsculas e minúsculas
 - Datas armazenadas como texto
 - Valores monetários inconsistentes
 - Campos obrigatórios vazios

O objetivo será criar uma base limpa e pronta para análise.

## 1. Importação da base

```python
# Bibliotecas

import pandas as pd
import numpy as np
```

```python
# Leitura do CSV

df = pd.read_csv(
    "vendas.csv"
)

df.head()
```

## 4. Tratamento de valores ausentes
### Cliente sem identificação

O campo cliente é importante para análises futuras.

Como não sabemos quem realizou o pedido, vamos substituir por:

Cliente Não Identificado

```python
df["cliente"] = df["cliente"].fillna(
    "Cliente Não Identificado"
)

df
```

### Forma de pagamento ausente

Neste caso o pedido está cancelado.

Vamos preencher:

Não informado

```python
df["forma_pagamento"] = df["forma_pagamento"].fillna(
    "Não informado"
)

df
```

### Produto ausente

```python
df["produto"] = df["produto"].fillna(
    "Produto Não Informado"
)
```

#### Tratar quantidade vazia

```python
df = df.dropna(
    subset=["quantidade"]
)
```

## 5. Padronização de textos

```python
df["cidade"] = df["cidade"].str.upper()
df["estado"] = df["estado"].str.upper()
df["status"] = df["status"].str.upper()
```

## 6. Removendo espaços extras

```python
df["cidade"] = df["cidade"].str.strip()
df["cliente"] = df["cliente"].str.strip()
```

## 7. Padronizando nomes das cidades

```python
df["cidade"] = (
    df["cidade"]
    .replace(
        {
            "SAO PAULO":"SÃO PAULO"
        }
    )
)
```

## 8. Conversão de datas

```python
df["data_pedido"] = pd.to_datetime(
    df["data_pedido"],
    dayfirst=True
)
```

## 9. Tratamento do valor unitário

```python
df["valor_unitario"]
```

```python
df["valor_unitario"] = (
    df["valor_unitario"]
    .astype(str)
    .str.replace(",",".")
)
```

```python
df["valor_unitario"] = pd.to_numeric(
    df["valor_unitario"],
    errors="coerce"
)
```

## 10. Criando uma coluna de faturamento

```python
df["valor_total"] = (
    df["quantidade"] *
    df["valor_unitario"]
)
```

```python
# Arredondamento
df["valor_total"] = df["valor_total"].round(2)
```

```python
# Visualizando
df[
[
"id_pedido",
"produto",
"valor_total"
]
]
```

## 11. Tratando status dos pedidos

```python
df["status"].value_counts()
```

## 12. Removendo pedidos cancelados

```python
df_vendas = df[
    df["status"]=="ENTREGUE"
]
```

## 14. Análise rápida após tratamento

#### Faturamento por cidade

```python
df_vendas.groupby(
    "cidade"
)["valor_total"].sum()
```

#### Produtos mais vendidos

```python
df_vendas.groupby(
    "produto"
)["quantidade"].sum().sort_values(
    ascending=False
)
```

```python
df_vendas["forma_pagamento"].value_counts()
```
